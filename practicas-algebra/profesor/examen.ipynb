```python
# Completar aquí
import numpy as np

A = np.array([
    [3., -4.], 
    [7., -6.]
])
print(f"A = \n {A}")
print(f"tipo de datos en A = {A.dtype}")

B = np.array([
    [7., 1.], 
    [0., -7.]
])
print(f"B = \n {B}")
print(f"tipo de datos en B = {B.dtype}")
# Fin Completar aquí ------------------------------------
```

```python
# Completar aquí
C = np.vstack((A, B))
print(f"C = \n {C}")
# Fin Completar aquí ------------------------------------
```

```python
# Completar aquí
from numpy.linalg import qr

Q, R = qr(C)

print(f"Q = \n {Q}")
print(f"R = \n {R}")
# Fin Completar aquí ------------------------------------
```

```python
# Completar aquí
Q_1 = Q[:2, :]
print(f"Q_1 = \n {Q_1}")

Q_2 = Q[2:, :]
print(f"Q_2 = \n {Q_2}")
# Fin Completar aquí ------------------------------------
```

```python
from scipy.linalg import svd

U_1, s_1, V_1_T = svd(Q_1)

# creamos una matriz de zeros de tamaño m x n 
Sigma_1 = np.zeros((Q_1.shape[0], Q_1.shape[0]))

# rellenamos S con los valores singulares
Sigma_1[:Q_1.shape[0], :Q_1.shape[0]] = np.diag(s_1)

print(f"tamaño de U_1 = {U_1.shape}")
print(f"tamaño de  Sigma_1 = {Sigma_1.shape}")
print(f"tamaño V_1^T = {V_1_T.shape}")
print(f"valores singulares de Q_1 = {s_1}")
print(f"U_1 = \n {U_1}")
print(f"V_1^T = \n {V_1_T}")
print(f"U_1 S_1 V_1^T = \n {U_1.dot(Sigma_1.dot(V_1_T))}")
# Fin Completar aquí ------------------------------------
```
