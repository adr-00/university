### Práctica 10: Compresión de imagenes vía SVD

```python
import numpy as np
```

```python
M = np.load("../data/beltrami_adult.npy")

print(f"La forma del array es: {np.shape(M)}")
```

```python
from PIL import Image

beltrami = Image.fromarray(M)
beltrami.show()
```

```python
A = M.T
```

```python
from scipy.linalg import svd

U, s, V = svd(A)   # La V es ya la traspuesta
Sigma  = np.zeros((A.shape[0], A.shape[1]))
Sigma[:A.shape[0], :A.shape[0]] = np.diag(s)


```

```python
r = 50

U_r = U[:, :r]
Sigma_r = Sigma[:r, :r]
V_r = V[:r, :]

print(f"La forma de U_r = {U_r.shape}")
print(f"La forma de Sigma_r = {Sigma_r.shape}")
print(f"La forma de V_r = {V_r.shape}")
```

```python
A_r = U_r.dot(Sigma_r.dot(V_r))
print(A_r.shape)
```

```python
M_r = A_r.T
```

```python
beltrami_compressed = Image.fromarray(M_r)
beltrami_compressed.show() 
```
