### Práctica 8: Métodos iterativos de resolución de sistemas lineales

#### Método de Jacobi

```python
import numpy as np
```

```python
A = np.array([
    [8, 3, -3],
    [-2, -8, 5],
    [3, 5, 10]
    ])
b = np.array([14, 5, -8])
```

```python
from metodos_iterativos import jacobi
```

```python
x = jacobi(A, b)
print(f"x = {x}")
```

#### Método de Gauss-Seidel

```python
from metodos_iterativos import gauss_seidell
```

```python
x = gauss_seidell(A, b)
print(f"x = {x}")
```

Veamos si la matriz es diagonal dominante

```python
diag = np.diag(np.abs(A))
off_diag = np.sum(np.abs(A), axis=1) - diag ###(0 = filas, 1 = columnas)
if np.all(diag > off_diag):
    print("Matriz diagonal dominante")
else:
    print("Matriz no diagonal dominante")
```
