### Introducción a Numpy n-dimensional arrays

[N-dimensional array](https://numpy.org/doc/stable/reference/generated/numpy.ndarray.html#numpy.ndarray)

```python
import numpy as np
```

```python
x = np.array([1 , 4, 3])
print(f"x = {x}")
print(type(x))
print(x.shape)
print(x.size)
print(x.dtype)
```

```python
A = np.array([[1, 4, 3],
             [9, 2, 7]])
print(f"A = \n {A}")
print(type(A))
print(A.shape)
print(A.size)
print(A.dtype)
```

#### Algunos arrays destacados

```python
A = np.zeros([3, 2])
print(f"A = \n {A}")
```

```python
c = np.arange(1, 5, 0.5)
print(c)
```

```python
array_complex = 1j * np.arange(5)
print(array_complex)
```

```python
z = np.ones(4, dtype=complex) * 2
print(z)
```

```python
d = np.linspace(0, 1, 10)
print(d)
```

Apilamiento de datos

```python
a1 = np.array([1, 2, 3])
a2 = np.array([4, 5, 6])
print(a1)
print(a2)
a3 = np.vstack((a1, a2))
print(a3)
a4 = np.hstack((a1, a2))
print(a4)
```

Listas, tuplas y arrays de numpy

```python
lista = [1, 2, 3]
print(lista)
print(type(lista))

vector = np.array(lista)
print(vector)
print(type(vector))
```

```python
data = [[11, 22],
        [33, 44],
        [55, 66]]
print(data)
print(type(data))

matriz = np.array(data)
print(data)
print(type(matriz))
```

```python
tupla = (1, 2, 3)
print(tupla)
print(type(tupla))

vector = np.array(tupla)
print(vector)
print(type(vector))
```
