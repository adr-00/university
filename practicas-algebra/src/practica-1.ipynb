### Segunda práctica

```python
def add_and_substract(iteration):
    result = 1

    for i in range(iteration):
        result += 1 / 3
    for i in range(iteration):
        result -= 1 / 3
    return result

add_and_substract(1000)

```

# Números complejos

```python
import cmath
```

```python
x = 1.
y = 1.
z = complex(x, y)
print(z)
print(type(z))
print(f"La parte real de z = {z.real}")
print(f"La parte real de z = {z.imag}")
```

```python
w = cmath.polar(z)
print(f"Módulo y argumento de z = {w}")
```

```python
import numpy as np
```

```python
print(f"Argumento o fase en grados = {np.rad2deg(w[1])}")
```

```python
print(f"Forma binómica de w = {cmath.rect(1.4142135623730951, 0.7853981633974483)}")
```

### Operaciones con complejos

```python
c1 = 1 + 1j
c2 = 0 + 0.5j
print(f"c1 + c2 = {c1 + c2}")
print(f"c1 - c2 = {c1 - c2}")
print(f"c1 * c2 = {c1 * c2}")
print(f"c1 / c2 = {c1 / c2}")
print(f"Cojugado de c1 = {c1.conjugate()}")
print(f"Potencia 4 de c2 = {c2 ** 4}")
```

#### Exponencial compleja

```python
z = complex(0, cmath.pi) #Número complejo binómica
print(f"Exponencial de z = {7 * cmath.exp(z)}") #Forma exponencial
```

#### Gráficas que involucarn números complejos

```python
import matplotlib.pyplot as plt
```

```python
#Ejercio hacho en la clase
t = np.linspace(0, 2 * np.pi, 1000)
f = np.exp(-t + 1j * 10 * t)
```

```python
x = np.real(f)
y = np.imag(f)
```

```python
fig, ax = plt.subplots()       #fig es como el folio, ax es como una matriz, subplots para más gráficos
ax.plot(x, y)
ax.set_title("Gráfica de $f(t)=e^{-t + j 10t}$")
plt.show

```

```python
fig, ax = plt.subplots()       #fig es como el folio, ax es como una matriz, subplots para más gráficos
ax.plot(t, x, label = "Parte real de $f$")
ax.plot(t, y, label = "Parte imaginaria de $f$")
ax.legend()
plt.show
```
