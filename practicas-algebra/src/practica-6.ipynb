```python
from scipy.sparse import csr_matrix
```

```python
A = np.array([
    [1, 0, 0, 1, 0, 0],
    [0, 0, 2, 0, 0, 1],
    [0, 0, 0, 2, 0, 0]
])
print(f"A = \n {A}")

S = csr_matrix(A)
print(f"A en formato sparse = {S}")

B = S.todense()
print(f"B = \n {B}")
```

Veamos como introducir una matriz directamente a sparse

```python
row = np.array([0, 0, 1, 1, 2])
col = np.array([0, 3, 2, 5, 3])
data = np.array([1, 1, 2, 1, 2])

mtx = csr_matrix((data, (row, col)), shape=(3, 6))
#print(f"Sparse matrix en formato csr = \n {mtx}")
#print(f"Misma matriz en formato dense = \n {mtx.todense()}")
print(f"Datos no nulos en la matriz en mtx = \n {mtx.data}")
print(f"Índices en mtx = \n {mtx.indices}")
print(f"Punteros para los índices de datos y filas = \n {mtx.indptr}")
```

```python
from scipy.sparse import csc_matrix
```

```python
A = np.array([
    [1, 0, 0, 1, 0, 0],
    [0, 0, 2, 0, 0, 1],
    [0, 0, 0, 2, 0, 0]
])
print(f"A = \n {A}")

S = csc_matrix(A)
print(f"Matriz en formato sparse por columnas = \n {S}")
```

Concepto de sparsity (Medida del número  de elementos no nulos)


```python
def sparsity(A):
    sparsi = 1.0 - np.count_nonzero(A) / A.size
    return sparsi

sparsity(A)
print(f"Sparsity de A = {sparsity(A)}")
```
