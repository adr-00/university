### Factorizaciones LU y PLU

```python
import numpy as np
```

```python
A = np.array([
    [4, -3, -5],
    [-2, -4, 5],
    [8, 8, 0]
])
print(f"A = \n {A}")
```

```python
from scipy.linalg import lu
```

```python
P , L , U = lu(A)
print(f"P = \n {P}")
print(f"L = \n {L}")
print(f"U = \n {U}")
```

```python
print(f"LU = \n {np.dot(L, U)}")
LU = np.dot(L, U) 
print(f"PLU = \n {np.dot(P, LU)}")
```
