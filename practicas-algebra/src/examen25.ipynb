```python
from scipy.sparse import csr_matrix
row = np.array([0, 1, 2, 3, 3, 3])
col = np.array([0, 1, 2, 2, 1, 0])
data = np.array([1, 1, 1, 1, 1, 1])

M_sparse = csr_matrix((data, (row, col)), shape=(4, 4))
M_dense = M_sparse.todense()

print(f"Matriz Z en formato sparse = \n {M_sparse}") 
print(f"Matriz Z en formato dense = \n {M_dense}")
```

```python
from scipy.linalg import svd

M = M_dense
U, s, V = svd(M)

Sigma = np.zeros((M.shape[0], M.shape[1]))

Sigma[:M.shape[0], :M.shape[1]] = np.diag(s)

print(f"U = \n {U}")
print(f"Sigma = \n {Sigma}")
print(f"V^T = {V}")
```

```python
u_1 = U[:, 0]
u_2 = U[:, 1]
u_3 = U[:, 2]
u_4 = U[:, 3]
print(f"La norma euclídea del vector columna u_1 = {np.linalg.norm(u_1, 2)}")
print(f"La norma euclídea del vector columna u_2 = {np.linalg.norm(u_2, 2)}")
print(f"La norma euclídea del vector columna u_3 = {np.linalg.norm(u_3, 2)}")
print(f"La norma euclídea del vector columna u_4 = {np.linalg.norm(u_4, 2)}")
print(f"El producto escalar del vector columa u_1 y u_2 = {u_1 @ u_2}")
print(f"El producto escalar del vector columa u_1 y u_3 = {u_1 @ u_3}")
print(f"El producto escalar del vector columa u_1 y u_4 = {u_1 @ u_4}")
print(f"El producto escalar del vector columa u_2 y u_3 = {u_2 @ u_3}")
print(f"El producto escalar del vector columa u_2 y u_4 = {u_2 @ u_4}")
print(f"El producto escalar del vector columa u_3 y u_4 = {u_3 @ u_4}")

V_inv = np.linalg.inv(V)
V_T = V.T

print(f"V^T == V^-1 ? {np.allclose(V_inv, V_T)}")
```

```python
r = np.linalg.matrix_rank(M)
print(f"El rango de M = {r}")
U_r = U[:, :r]

Sigma_r = Sigma[:r, :r]

V_r = V[:r, :]
print(f"U_r = {U_r}")
print(f"Sigma_r = {Sigma_r}")
print(f"V_r = {V_r}")
print(np.allclose(M, U_r @ Sigma_r @ V_r))
u_1 = U_r[:, 0]
u_2 = U_r[:, 1]
u_3 = U_r[:, 2]
v_1 = V_r[0, :]
v_2 = V_r[1, :]
v_3 = V_r[2, :]

print(f"Descomposicion espectral = {Sigma_r[0,0]} * {(u_1.reshape(4, 1) @ v_1.reshape(1, 4))} \n + {Sigma_r[1,1]} * {(u_2.reshape(4, 1) @ v_2.reshape(1, 4))} \n + {Sigma_r[2,2]} * {(u_3.reshape(4, 1) @ v_3.reshape(1, 4))}")
```
