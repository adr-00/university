## Práctica 4: Repasar vectores y matrices: aritmética y normas.

### Vectores

```python
import numpy as np
```

```python
v = np.array([1, 2, 3])
print(f"v = {v}")

u = np.array([1, 2, 3])
print(f"u = {u}")

#Suma y resta de vectores
print(f"v + u = {v + u}")
print(f"v - u = {v - u}")

#Producto y cociente componente a componente
print(f"u * v = {u * v}")
print(f"u / v = {u / v}")
```

```python
#Escalar por un vector
a = 2.0
print(f"a * v = {a * v}")
```

```python
#Producto escalar
print(f"u.v = {u.dot(v)}")
```

```python
#Normas de vectores
l2 = np.linalg.norm(u, 2)
print(f"La norma 2 de u = {l2}")

l1 = np.linalg.norm(u, 1)
print(f"La norma 2 de u = {l1}")

lmax = np.linalg.norm(u, np.inf)
print(f"La norma 2 de u = {lmax}")

```

### Matrices

```python
A = np.array([
    [1, 3, 5],
    [7, 9, 11]
])
print(f"A = \n {A}")

B = np.array([
    [0, 2, 4],
    [6, 8, 10]
])
print(f"B = \n {B}")

#Suma y resta de matrices
print(f"A + B = \n {A + B}")
print(f"A - B = \n {A - B}")

#Producto de Hadamard (componente a componente), con el mismo tamaño
print(f"A * B = \n {A * B}")
print(f"A / B = \n {B / A}")

print(f"A.B = \n {A.dot(B)}")

```

```python
#Producto de matrices

A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

B = np.array([
    [1, 2],
    [3, 4]
])

print(f"A.B = \n {A.dot(B)}") #Producto usual
print(f"A.B = \n {A @ B}") #Otra forma
```

```python
#Producto matriz por vector
b = np.array([0.5, 0.5]) #EL vector es vertical
print(b)
print(A)
print(f"A.b = {A.dot(b)}")
```

```python
b = 0.5
print(f"b * A = \n {b * A}")
```

```python
#Algunos tipos de matrices
A = np.array([
    [2, -1, 0],
    [-1, 2, -1],
    [0, -1, 2]
])
print(f"A = \n {A}")

print(f"A^T = \n {A.T}")
# ¿Es A Simétrica?
A == A.T
```

```python
#Matrices diagonales
d = np.diag(A)
print(f"La diagonal de A = \n {d}")

D = np.diag(d)
print(f"D = \n {D}")
```

```python
#Matriz identidad
I = np.identity(4)
print(f"I = \n {I}")
```

```python
#Inversa, traza, determinante y rango de una matriz
B = np.linalg.inv(A)  ## Inversa
print(f"La inversa de A = \n {B}")
print(f"A.B = \n {B @ A}")


print(f"La traza de A = {np.trace(A)}") ##Traza

print(f"El determinante de A = {np.linalg.det(A)}") ##Determinante

print(f"El rango de A = {np.linalg.matrix_rank(A)}")  ##Rango
```
