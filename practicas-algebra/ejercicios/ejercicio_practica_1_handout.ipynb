Consider los números complejos $z_1 = a + bj$ y $z_2 = g + hj$.
Calcula la forma polar de ambos números. Expresa la fase en grados

```python
# Completar aquí
import cmath
import numpy as np
a = 2
b = 9
g = 7
h = 9

z1 = complex(a, b)
z2 = complex(g, h)

polar_z1 = cmath.polar(z1)
polar_z2 = cmath.polar(z2)
print(f"z_1 en forma polar es {polar_z1}")
print(f"Argumento o fase en grados = {np.rad2deg(polar_z1[1])}")
print(f"z_2 en forma polar es {polar_z2}")
print(f"Argumento o fase en grados = {np.rad2deg(polar_z2[1])}")
# --------------------
```

```python
import matplotlib.pyplot as plt

t = np.linspace(0, 4 * cmath.pi, 10000)
f = (1 / (1 + t)) * np.exp(np.pi * 1j + 1j * t )
x = np.real(f)
y = np.imag(f)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title(r"Gráfica de $f(t) = j1/(1 + t)e^{ j (t + \pi/2)}, \quad 0\leq t\leq 4\pi .$")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(t, x, label = "Parte real de $f$")
ax.plot(t, y, label = "Parte imaginaria de $f$")
ax.legend()
plt.show
```
