```python
import numpy as np
import matplotlib.pyplot as plt

# Definimos una sucesión acotada
n = np.arange(1, 100)
a_n = np.sin(n / 10)  # Sucesión acotada entre -1 y 1

# Seleccionamos una subsecuencia que converge a un límite
subsequence_indices = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
subsequence = a_n[subsequence_indices]

# Límite al que converge la subsecuencia (ejemplo: 0)
limit = 0

# Graficamos la sucesión
plt.figure(figsize=(10, 6))
plt.plot(n, a_n, label='Sucesión $a_n = \sin(n/10)$', marker='o', markersize=4, linestyle='-', color='lightblue')
plt.plot(subsequence_indices, subsequence, label='Subsecuencia Convergente', marker='o', markersize=8, color='orange')
plt.axhline(y=limit, color='red', linestyle='--', label='Límite (0)')

# Configuración del gráfico
plt.title('Teorema de Bolzano-Weierstrass en Sucesiones')
plt.xlabel('n')
plt.ylabel('$a_n$')
plt.legend()
plt.grid()
plt.ylim(-1.5, 1.5)

# Mostrar el gráfico
plt.show()

```
