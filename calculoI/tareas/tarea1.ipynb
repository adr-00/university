#### Ejercicio 1

Apartado 1

```python
##Sucesión de la suma de los divisores de un número natural

def suma_divisores(n):
    #Creamos una lista para almacenar los divisores
    x = []
    #Creamos un bucle que vaya sacando los divisores de "n", 
    #a tráves de la operación módulo y los añada a la lista.
    for i in range(1, n + 1): 
        if n % i == 0:
            x.append(i)
    #Se suman los términos de la lista
    return(sum(x))
```

Apartado 2

```python
#Sumatorio de 1/k
def armonico(n):
    #Creamos una variable donde se vaya guardando la suma
    suma = 0
    #Creamos un bucle que vaya sumando el resultado de 1/i,
    #donde i va tomando valores hasta n
    for i in range(1, n + 1):
        suma = suma + (1/i)
    #Nos devuelve la suma
    return(suma)
```

Apartado 3

```python
def R(n):
    result = armonico(n) + math.log(armonico(n)) * math.exp(armonico(n))
    return(result)
```

Apartado 4

```python

plt.figure(figsize=(10, 6))
n = 1 + np.arange(100)
plt.plot(n, [suma_divisores(i) for i in n], linestyle='-', color='b', label='Suma de Divisores')
plt.plot(n, [R(i) for i in n], linestyle='--', color='g', label='R(n)')
plt.title("Podemos observar que d(n) ≤ R(n)")
plt.legend()
plt.show()
```

```python
#Declaramos las variables que nos indicaran la posición del contador y 
#el número de números que cumple la desigualdad
numeros = 0
contador = 1

#Creamos un bucle hasta que haya 4 numeros que cumplan la condición
while numeros < 4:
    division = (R(contador)/suma_divisores(contador))
    #Comprobamos que la división cumple la desigualdad, si la cumple la mostramos
    #en pantalla y sumamos uno a la variable números.
    if division < (1 + 0.02):
        print(f"La posición {contador} con el número {division}")
        numeros += 1
    #Cada vez aumentamos el contador 1
    contador += 1




    
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Creamos la función zeta
def zeta(s, N):
    sum = 0
    for n in range(1, N+1):
        sum += 1 / (n**s)
    return sum

# Parámetros
x_values = np.linspace(1.01, 5, 100)
N = 1000  # Número de términos a sumar

# Calcular los valores de la función zeta
y = [zeta(s, N) for s in x_values]

plt.plot(x_values, y)
plt.xlabel('x')
plt.ylabel('ζ(s)')
plt.title('Aproximación de la función Zeta de Riemann')
plt.grid(True)
plt.show()
```

```python
#Fórmula de donde se sacan los despejes:
# pendiente = (y2 - y1)/(x2 - x1)
def pendiente(a, b, c):
    # Comprobamos que c sea mayor o menor de 0.5, en el caso de 0.5 nos daría igual
    if c <= 0.5:
        # La base tiene el valor de (c-1)
        # Calculamos la altura del punto a averiguar (x,y), en este caso la y
        altura = (b) / (2*(1-c))
        # Calculamos la pendiente que corta con con el punto (x,y) a averiguar    
        pendiente_corta = b / (a-1)
        # Con la pendiente sacamos la x
        x = (altura + pendiente_corta) / pendiente_corta
        # Apartir del punto (x,y), sacamos la pendiente que nos pide el enunciado
        if x == c :
            # Si x == c, nos da error por que se divide entre cero, lo que significa
            #que es perpendicular
            print("La pendiente es perpendicular")
        else:
            pendiente = altura / (x-c)
            print(f"La pendiente es {pendiente}")
    elif c > 0.5:
        # En este caso la la base tiene el valor de c
        altura = (b) / (2*c)
        pendiente_corta = b / a
        x = altura / pendiente_corta
        if x == c:
            print("La pendiente es perpendicular")
        else:
            pendiente = altura / (x-c)
            print(f"La pendiente es {pendiente}")
    return x, altura
pendiente(a, b, c)
```

```python
x, altura = pendiente(a, b, c)
x_coords = [0, 1, a, 0]
y_coords = [0, 0, b, 0]
    
plt.plot(x_coords, y_coords, 'g-', marker='o') 
plt.fill(x_coords, y_coords, 'lightgreen', alpha=0.5)
plt.plot(x, altura, "b", marker='o')
plt.plot(c, 0, "b", marker='o')
plt.plot((x, c),(altura, 0))

plt.grid(True)
plt.show()
```
