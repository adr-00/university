#### Representación de funciones en 3D

```python
from numpy import linspace, meshgrid
x = linspace(-1.5, 1.5, 200)
y = linspace(-1.5, 1.5, 200)
X, Y = meshgrid(x, y)
Z=X**2+Y**2
```

```python
from matplotlib.pyplot import subplots
fig, ax = subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z)
```

#### Representar un círculo, coordenadas polares

```python
from numpy import linspace, newaxis, pi
r = linspace(0.125, 1.0, 8) #empiezo en un itervalo inferior a 0 ya que en cero se anula
ang = linspace(0, 2*pi, 36, endpoint=False)[..., newaxis] # sin el 2pi
```

```python
# Siempre se pinta en coordenadadas cartesianas
from numpy import sin, cos, append
x = append(0, (r*cos(ang)).flatten())
y = append(0, (r*sin(ang)).flatten())
z = sin(-x*y)
```

```python
from matplotlib.pyplot import subplots
fig, ax = subplots(subplot_kw={"projection": "3d"})
ax.plot_trisurf(x, y, z)  # PAra pintar se utiliza el trisurf
```

#### Como pintar varias gráficas a la vez

```python
from numpy import linspace, meshgrid
f = lambda x, y: x**2 + y**3 + 3*x*y
p = lambda x, y: 5*x + 6*y-6
x = linspace(-2, 4, 50)
y = linspace(-2, 4, 50)
X, Y = meshgrid(x, y)
```

```python
from matplotlib.pyplot import figure, show
fig = figure(figsize=(7, 6))
ax = fig.add_axes([0, 0, 1, 1], projection="3d")
ax.plot_surface(X, Y, f(X, Y), color="yellow", alpha=0.5)
ax.plot_surface(X, Y, p(X, Y), color="blue", alpha=0.25)
show()
```
