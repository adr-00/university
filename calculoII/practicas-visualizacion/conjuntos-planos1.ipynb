#### 2.2.2. Visualización de conjuntos planos

```python
from numpy import linspace, meshgrid
from matplotlib.pyplot import contour, contourf
```

```python
x = linspace(-1.5, 1.5, 200)
y = linspace(-1.5, 1.5, 200)
X, Y = meshgrid(x, y)
f = lambda x, y: x**2 + y**2
```

```python
contour(X, Y, f(X, Y), [1], colors='green')
contour(X, Y, f(X, Y), [2], colors='red')
```

```python
contourf(X, Y, f(X, Y), [1, 'inf'], colors='blue', alpha=0.5)
# El alfa se utiliza para la opacidad
```

Voy a representar un conjunto

```python
#Lo que no este pintado va a ser mi conjunto, importamos figure para coloear varios graficos
from matplotlib.pyplot import contourf, figure
f1 = lambda x, y: x**2 + y**2
f2 = lambda x, y: x
figure(figsize=(5, 5))
contourf(X, Y, f1(X, Y), [1, 'inf'], colors='blue')
contourf(X, Y, f2(X, Y), ['-inf',0], colors='green')
```

Si quiero que los bordes entren

```python

figure(figsize=(5, 5))
contourf(X, Y, f1(X, Y), [1, 'inf'], colors='blue')
contourf(X, Y, f2(X, Y), ['-inf',0], colors='green')
contour(X, Y, f1(X, Y), [1], colors='black')
```
