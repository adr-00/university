#### Gráficas con matplotlib

```python
from numpy import linspace, sin, pi
from matplotlib.pyplot import subplots
```

```python
x = linspace(-pi, pi, 200)
y = sin(x**2 + 1)
```

```python
fig, ax = subplots()
ax.plot(x, y)
```

##### Ejercicio 87

```python
x = linspace(-2*pi, +2*pi, 200)
f = sin(x**2)
g = (sin(x))**2

fig, ((ax1), (ax2)) = subplots(2, 1)
ax1.plot(x, f)
ax2.plot(x, g)
```

```python
x = linspace(-2*pi, +2*pi, 200)
f = sin(x**2)
g = (sin(x))**2

fig, (ax1, ax2) = subplots(1, 2)
ax1.plot(x, f)
ax2.plot(x, g)
```

```python
x = linspace(-2*pi, +2*pi, 200)
f = sin(x**2)
g = (sin(x**2))**2

fig, ((ax1), (ax2)) = subplots(2, 1)
ax1.plot(x, f)
ax2.plot(x, g)
```

```python
x = linspace(-2, +2, 200)
f1 = x**2 + 1
f2 = x**3 + 1
f3 = x**2 - 1
f4 = (sin(x**2))**2

fig, ((ax1, ax2), (ax3, ax4)) = subplots(2, 2)
ax1.plot(x, f1)
ax2.plot(x, f2)
ax3.plot(x, f3)
ax4.plot(x, f4)
```

```python
from numpy import meshgrid  # Coge la coordenada x y la y me da todos los puntos
from matplotlib.pyplot import contour, contourf
x = linspace(-15, 15, 200)
y = linspace(-15, 15, 200)
X, Y = meshgrid(x, y)

f = lambda x, y: x**3 + 2*x*y + y**3
contour(X, Y, f(X, Y), [1], colors='blue')
```
