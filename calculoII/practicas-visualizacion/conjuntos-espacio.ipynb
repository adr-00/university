## Visualización de conjuntos en el espacio

```python
from numpy import linspace, meshgrid
f1 = lambda x, y: 4 - x**2 - x*y
f2 = lambda x, y: x + y
x = linspace(0, 1, 20)
y = linspace(0, 1, 20)
X, Y = meshgrid(x, y)
```

```python
from matplotlib.pyplot import figure, show
fig = figure(figsize=(7, 6))
ax = fig.add_axes([0, 0, 1, 1], projection="3d")
ax.plot_surface(X, Y, f1(X, Y), color="green", alpha=0.5)
ax.plot_surface(X, Y, f2(X, Y), color="blue", alpha=0.5)
show()
```

Si iguales las dos fucniones para ver donde son iguales, nos sale una circunferencia de radio 1

```python
from numpy import linspace, newaxis, pi
r = linspace(0.125, 1.0, 8)
ang = linspace(0, 2*pi, 36, endpoint=False)[..., newaxis]
```

```python
from numpy import sin, cos, append
x = append(0, (r*cos(ang)).flatten())
y = append(0, (r*sin(ang)).flatten())
z1 = x**2+y**2
z2 = 2-x**2-y**2
z3=0.000000001*x
```

```python
from matplotlib.pyplot import figure, show
fig = figure(figsize=(7, 6))
ax = fig.add_axes([0, 0, 1, 1], projection="3d")
ax.plot_trisurf(x, y, z1, color="blue", alpha=0.5)
ax.plot_trisurf(x, y, z2, color="yellow", alpha=0.5)
ax.plot_trisurf(x, y, z3, color="gray", alpha=0.5)
show()
```

#### Pintamos una esfera (No va)

```python
import numpy as np
from matplotlib.pyplot import figure, show

# Crear una cuadrícula de puntos en coordenadas esféricas
phi = np.linspace(0, 2 * np.pi, 50)  # Ángulo azimutal
theta = np.linspace(0, np.pi, 50)   # Ángulo polar
phi, theta = np.meshgrid(phi, theta)

# Convertir a coordenadas cartesianas
r = 1  # Radio de la esfera
x = r * np.sin(theta) * np.cos(phi)
y = r * np.sin(theta) * np.sin(phi)
z = r * np.cos(theta)

# Graficar la esfera
fig = figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(x, y, z, color="blue", alpha=0.6)  # Superficie de la esfera
show()
```

#### Cilindro

```python
from numpy import linspace, meshgrid, sqrt
x = linspace(-1, 1, 50)
y = linspace(-1, 1, 50)
X, Y = meshgrid(x, y)
Z1 = sqrt(1-X**2)
Z2 = -sqrt(1-X**2)
```

```python
from matplotlib.pyplot import figure, show
fig = figure(figsize=(7, 6))
ax = fig.add_axes([0, 0, 1, 1], projection="3d")
ax.plot_surface(X, Y, Z1, color="green", alpha=0.5)
ax.plot_surface(X, Y, Z2, color="blue", alpha=0.5)
show()
```
