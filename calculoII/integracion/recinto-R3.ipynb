```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Configuración de la figura 3D
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Parámetros para la malla
x = np.linspace(-1, 0.5, 100)
y = np.linspace(-0.8, 0.8, 100)
x, y = np.meshgrid(x, y)

# Cilindro parabólico superior: z = 1 - y^2
z_superior = 1 - y**2
ax.plot_surface(x, y, z_superior, alpha=0.5, color='blue', label='Cilindro parabólico $z=1-y^2$')

# Plano inferior: z = -2x -3y -10
z_inferior = -2*x - 3*y - 10
ax.plot_surface(x, y, z_inferior, alpha=0.5, color='red', label='Plano $z=-2x-3y-10$')

# Cilindro circular lateral: x^2 + y^2 + x = 0 (completar cuadrados)
theta = np.linspace(0, 2*np.pi, 100)
r = 0.5  # Radio del cilindro
x_centro = -0.5  # Centro en x
x_cilindro = x_centro + r * np.cos(theta)
y_cilindro = r * np.sin(theta)
z_cilindro = np.linspace(-15, 2, 100)  # Rango de z para el cilindro
X_cil, Z_cil = np.meshgrid(x_cilindro, z_cilindro)
Y_cil, Z_cil = np.meshgrid(y_cilindro, z_cilindro)
ax.plot_surface(X_cil, Y_cil, Z_cil, alpha=0.3, color='green', label='Cilindro lateral $x^2+y^2+x=0$')

# Ajustes visuales
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Sólido limitado por superficies dadas')
ax.legend()
plt.tight_layout()
plt.show()
```
