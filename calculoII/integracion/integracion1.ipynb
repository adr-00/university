```python
import sympy as sy
import numpy as np
```

```python
x,y = sy.symbols('x y') # Solo en el sympy
```

```python
u = np.linspace(-1, 4, 200)
v = np.linspace(-2, 2, 200)
U, V = np.meshgrid(u, v)
```

```python
f = lambda u, v: u+4*v**2 - 3
g = lambda u, v: u 
```

```python
from matplotlib.pyplot import contour, contourf, figure
figure(figsize=(5, 5))
contourf(U, V, f(U, V), [0, 'inf'], colors='blue')
contourf(U, V, g(U, V), ['-inf', 0], colors='blue')
```

```python
from numpy import linspace, meshgrid

x = linspace(-2, 5, 100)
y = linspace(-2, 5, 100)
X, Y = meshgrid(x, y)

f1 = lambda x, y: x**2 + y**2 -2*x
f2 = lambda x, y: x**2 + y**2 -4*x
f3 = lambda x, y: y - x
f4 = lambda x, y: y

contour(X, Y, f1(X, Y), [0], colors='green')
contour(X, Y, f2(X, Y), [0], colors='black')
contour(X, Y, f3(X, Y), [0], colors='black')
contour(X, Y, f4(X, Y), [0], colors='black')
```
