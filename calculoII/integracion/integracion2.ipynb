```python
from numpy import linspace, meshgrid

x = linspace(-2, 5, 100)
y = linspace(-2, 5, 100)
X, Y = meshgrid(x, y)

f1 = lambda x, y: x**2 + y**2 -2*x
f2 = lambda x, y: x**2 + y**2 -4*x
f3 = lambda x, y: y - x
f4 = lambda x, y: y
```

```python
from matplotlib.pyplot import contour, contourf
contour(X, Y, f3(X, Y), [0], colors='blue')
contour(X, Y, f4(X, Y), [0], colors='red')
contour(X, Y, f1(X, Y), [0], colors='green')
contour(X, Y, f2(X, Y), [0], colors='yellow')
```
