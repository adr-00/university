```python
from sympy import *
```

```python
X, Y, Z, z, t = symbols("X Y Z z t")
```

```python
solve([z*X+2*X+Z-(1/z),z*Y-X+2*Y,z*Z-2*X+4*Z+(2/z)],[X, Y, Z])
```

```python
f1 = inverse_laplace_transform((z + 2)/(z**3 + 6*z**2 + 10*z),z,t)
f2 = inverse_laplace_transform(1/(z**3 + 6*z**2 + 10*z),z,t)
f3 = inverse_laplace_transform((2*z + 6)/(z**3 + 6*z**2 + 10*z),z,t)
```

```python
print(f3)
```

```python
plot(f1, f2, f3, (t, 0, 10))
```
