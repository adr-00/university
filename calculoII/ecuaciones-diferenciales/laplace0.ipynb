## Laplace

```python
from sympy import symbols, laplace_transform, Heaviside, sin, exp, sinh, cos

t, z = symbols("t z")
laplace_transform(t**2+t+1,t,z)
```

```python
laplace_transform(Heaviside(t),t,z,noconds=True)
```

```python
laplace_transform(Heaviside(t - 2),t,z,noconds=True)
```

#### Ejercicio 2

```python
laplace_transform(t* (Heaviside(t) - Heaviside(t - 1)) + 1*(Heaviside(t-1) - Heaviside(t-2)),t,z,noconds=True)
```

### Inversa de Laplace

```python
from sympy import inverse_laplace_transform, pi
```

```python
inverse_laplace_transform(1/(1+z**2), z, t)
```

```python
inverse_laplace_transform((z**2+z+1)/(z**2+1)**2, z, t)
```

```python
inverse_laplace_transform(exp(-2*z)/(1+z**2),z,t)
```

#### Ejercicio 1

```python
inverse_laplace_transform(z**2/(1+z**3),z,t)
```

```python
inverse_laplace_transform(1/((z - 3)*(z**2 - 2)),z,t)
```

```python
inverse_laplace_transform((z - 7)/(z**2 + 2*z + 5),z,t)
```

```python
inverse_laplace_transform(1/((z + 1)*(z + 2)*(z**2 + 2*z + 10)),z,t)
```

#### Ejercicio 2

```python
inverse_laplace_transform((z*exp(-pi*z))/(z**2 + 2*z + 5),z,t)
```

```python
inverse_laplace_transform((z + 1)/(exp(z)*z**2*(z**2 + 9)), z, t)
```

```python
inverse_laplace_transform(((z + 1)/(z**4)),z,t)
```
