#### Antes de todo pasar la ecuación a Laplace

##### 4.3 primera ecuación

##### l es la tranformada de Laplace

```python
t, z, l = symbols('t z l')
```

```python
eq = (z**2 + 1)*l-z-1-(1/z**2 + 1)
```

```python
solve(eq, l)
```

```python
inverse_laplace_transform((z**3 + 2*z**2 + 1)/(z**4 + z**2), z, t)
```
