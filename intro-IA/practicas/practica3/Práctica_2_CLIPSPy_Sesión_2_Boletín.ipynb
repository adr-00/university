**Restricciones sobre el valor retornado en la llamada a una función.**

Además de las restricciones literales que se han visto en la sesión anterior, se pueden establecer restricciones sobre el valor retornado en la llamada a una función. Se denota por =.  

El valor devuelto por la función debe ser uno de los tipos de datos primitivos.

La función adición en CLIPS se realiza de la siguiente forma:
(+ <expresión-numérica>)
Ejemplo: (+ 1 2 3 4) da como resultado 10.

La función multiplicación en CLIPS se realiza de la siguiente forma:
(* <expresión-numérica>)
Ejemplo: (* 1 2 3 4) da como resultado 24.


La sustracción, la división, el valor absoluto, el máximo y el mínimo se definen de forma análoga.


```python
# Importación de la biblioteca CLIPSPy
import clips
```

```python
# Creamos el entorno
entorno_ejercicios = clips.Environment()
```

Regla que emplea la restricción =  y una variable unicampo

```python
# Regla que emplea la restricción =  y una variable unicampo
regla_ejemplo_restriccion_1 = """
(defrule ejemplo_restriccion_1 "Ejemplo de uso de ="
  ( dato ?x )
  ( Persona (edad = (* 2 ?x) ) )
  =>
  ( assert(Encontrada_persona_con_edad_igual_al_doble_de ?x) )
)
"""
```

Cuando la restricción = se emplea con una variable unicampo hay que escribir &=,  la regla emplea además dos variables unicampo (?x e ?y) y la conectiva de casilla |.

```python
# Regla que emplea la restricción =,  dos variables unicampo
# y conectivas de casilla & y |
regla_ejemplo_restriccion_2 = """
(defrule ejemplo_restriccion_2 "Ejemplo de uso de ="
  ( dato ?x )
  ( Persona (edad ?y &=(* 2 ?x) | =(+ 3 ?x)) )
  =>
  ( assert(Encontrada_persona_con_edad_igual_a ?y) )
  ( assert(Encontrada_persona_con_edad_doble de ?x o_3_años_más_que ?x) )
)
"""
```

**Restricciones predicado.**

En ocasiones es necesario restringir un campo basado en la veracidad de una expresión booleana. Esta restricción se realiza mediante la restricción predicado que se denota por **:** . El modo de usar esta restricción es como sigue:

?x & : <restricción-con-?x>

La expresión booleana con **valores numéricos** se puede realizar con las siguientes operaciones de comparación numérica:

•	**Igualdad**. (= <expresión-numérica> < expresión-numérica >+).
Retorna el valor TRUE si el primer argumento es igual en valor a todos los demás argumentos.

•	**Desigualdad**. (<> < expresión-numérica > < expresión-numérica >+).
Retorna el valor TRUE si el primer argumento no es igual en valor a todos los demás argumentos.

•	**Mayor que**. (>  < expresión-numérica > < expresión-numérica >+).
Retorna el valor TRUE si el argumento i es mayor que el argumento i + 1.

•	**Mayor o igual que**. (>=  < expresión-numérica > < expresión-numérica >+).
Retorna el valor TRUE si el argumento i es mayor o igual que el argumento i + 1.

•	**Menor que**. (<  <expresión-numérica > < expresión-numérica >+).
Retorna el valor TRUE si el argumento i es menor que el argumento i + 1.

•	**Menor o igual que**. (<=  < expresión-numérica > < expresión-numérica >+).
Retorna el valor TRUE si el argumento i es menor o igual que el argumento i + 1.

La expresión booleana puede ser la comparación de igualdad o desilguadad con **cualquier tipo de datos** de CLIPSPy gracias a las dos siguientes funciones:

•	**eq**. Retorna el valor TRUE si el primer argumento es igual en valor a todos los demás argumentos.

•	**neq**. Retorna el valor TRUE si el primer argumento no es igual en valor a todos los demás argumentos.


En la siguiente celda se muestra un ejemplo de regla de este tipo que hace uso de la operación mayor que.

```python
# Regla que emplea la restricción predicado
regla_ejemplo_restriccion_predicado = """
(defrule ejemplo_restriccion_predicado "Ejemplo de uso de :"
  (dato ?x & : (> ?x 0))
  =>
  ( assert(El_número ?x es_positivo) )
)
"""

# Generamos el hecho ordenado negativo para probar que la regla no se activa
# con este hecho
# La regla se activará con los hechos del ejercicio anterior
entorno_ejercicios.assert_string('(dato 0)')

# Añadimos la regla al entorno
entorno_ejercicios.build(regla_ejemplo_restriccion_predicado)

# Mostramos las reglas
for rule in entorno_ejercicios.rules():
  print(rule)

# Se ejecuta el motor de inferencia: comprueba las reglas que se actiavn y procede a ejecutarlas en orden
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)
```

**Elemento condicional test.**

El elemento condicional test permite evaluar expresiones en el antecedente de la regla. La evaluación puede resultar no-FALSE o FALSE:

•	Si es no-FALSE, la regla puede ser activada si también se satisfacen los restantes elementos condicionales de la regla.

•	Si es FALSE, la regla nunca se activará, salvo que existan elementos condicionales conectivos que hagan que el antecedente se satisfaga.

En la siguiente celda vemos cómo emplear el elemento condicional *test* para comprobar si un número es positivo, que se había programado anteriormente con la restricción de predicado **&:**

```python
# Regla que emplea el elemento condicional test para realizar la operación
# de comprobación de si un número es positivo. La conclusión se guarda como valor de un
# campo de un hecho
regla_comprobacion_numero_positivo = """
(defrule comprobacion_numero_positivo
  ( dato ?x )
  ( test (> ?x 0) )
  =>
  ( assert(El_número ?x es_positivo) )
)
"""

# Generamos el hecho dato necesario
entorno_ejercicios.assert_string('(dato 10)')

# Mostramos los hechos antes de la ejecución de la última regla
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)

# Añadimos la regla al entorno
entorno_ejercicios.build(regla_comprobacion_numero_positivo)

# Con el siguiente código podemos comprobar las reglas existentes en el entorno
for rule in entorno_ejercicios.rules():
    print(rule)

# Se ejecuta el motor de inferencia: comprueba las reglas que se actiavn y procede a ejecutarlas en orden
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)
```

Este elemento condicional también puede utilizarse con elementos condicionales conectivos. La siguiente celda muestra un ejemplo de este tipo. La siguiente regla comprueba si la diferencia entres dos números es menor que 2, que los dos números son positivos y que son diferentes

```python
# Regla que emplea el elemento condicional test y el elemento condicional conectivo and
# La regla comprueba si la diferencia entres dos números es menor que 2 y que los dos números son positivos
# y que son diferentes
regla_comprobacion_relacion_numeros = """
(defrule comprobacion_relacion_numeros
  ( dato_1 ?x )
  ( dato_2 ?y )
  ( test (and
        (> ?x 0)
        (> ?y 0)
        (< (abs(- ?x ?y)) 2)
        (<> ?x ?y)
        )
  )
  =>
  ( assert(Condicion se_cumple) )
)
"""
```

Comprobamos que la regla funciona:

```python
# Generamos el hecho dato necesario
entorno_ejercicios.assert_string('(dato_1 10)')
entorno_ejercicios.assert_string('(dato_2 11)')

# Mostramos los hechos antes de la ejecución de la última regla
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)

# Añadimos la regla al entorno
entorno_ejercicios.build(regla_comprobacion_relacion_numeros)

# Con el siguiente código podemos comprobar las reglas existentes en el entorno
for rule in entorno_ejercicios.rules():
    print(rule)

# Se ejecuta el motor de inferencia: comprueba las reglas que se actiavn y procede a ejecutarlas en orden
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)
```

## Resolución de conflictos entre reglas

La agenda de CLIPSPy contiene las reglas activadas candidatas a ser disparadas. La agenda funciona como una pila: la regla que se encuentra en el tope de la pila es la primera en ser ejecutada. Cuando una regla se añade a la agenda, se apila atendiendo a estos factores:


1. La regla nueva, que tendrá un valor de prioridad, se coloca por debajo de todas aquellas reglas que tiene una mayor prioridad y por encima de las que tiene una prioridad menor a la nueva regla.

2. Entre las reglas con igual prioridad, se utiliza la estrategia de resolución de conflictos para reordenarlas.

3. En caso de empate, tras aplicar los pasos anteriores, CLIPSPy colocaría de forma arbitraria las reglas (normalmente por el orden de definición de las reglas).

La prioridad (en ingles, *salience*) de una regla se indica con un valor numérico. Por defecto es 0, y está comprendido entre -10000 y +10000. El usuario puede asignar directamente dicho valor mediante una constante o una expresión numérica. La asignación se realiza escribiendo la siguiente expresión entre el nombre de la regla y el primer elemento condicional de la regla:


`(declare (salience <expresion-numerica>))`

Ya que salience es una herramienta muy potente para controlar la ejecución, podría pensar en usarla para establecer el orden de disparo de todas las reglas. Sin embargo, esta forma de programación en un sistema basado en reglas tiene limitaciones y presenta problemas:

<ul>
<li>Se estaría aplicando una técnica de programación procedural (procedimental) y que se indica el orden de ejecución de todas las reglas. CLIPS no es un lenguaje de programación adecuado para ese tipo de programación.</li>

<li>No se aprovecha las ventajas de la programación basada en reglas. La principal ventaja de la programación basada en reglas radica en que el programador no se preocupa de la ejecución del programa. Tan sólo se limita a codificar el conocimiento que se tiene de un problema para que, posteriormente, atendiendo a criterios naturales de ejecución, se disparen las reglas de forma óptima.</li>
</ul>





## Aplicación de CLIPS: sistema experto de reparación de un coche eléctrico.

Un conjunto de posibles reglas simples de un sistema experto en ofrecer recomendaciones sobre la reparación de un coche eléctrico es:

<ul>

<li> Si el coche arranca, entonces hay que preguntar si el coche funciona correctamente.</li>

<li> Si el coche no arranca, entonces hay que preguntar si la batería esta descargada.</li>

<li> Si la batería está descargada, entonces hay que conectar el coche a la red eléctrica para que cargue la batería. Fin del programa.</li>

<li> Si la batería no está descargada, entonces hay que preguntar si la prueba de conexión de la batería al motor es positiva.</li>

<li> Si la prueba de conexión de la batería al motor es positiva, entonces hay que llevar el coche al taller.</li>

<li> Si la prueba de conexión de la batería al motor es negativa, entonces la reparación sugerida es la de reparar la conexión de la batería y el motor.</li>

<li> Si el coche funciona correctamente, entonces no hay que repararlo ni llevarlo al taller. Fin del programa.</li>

<li> Si el coche no funciona correctamente, entonces hay que preguntar si el coche va lento.</li>

<li> Si el coche va lento, hay que preguntar si funciona el sistema de enfriamiento de las baterías.</li>

<li> Si el sistema de enfriamiento de las baterías no funciona, entonces la reparación sugerida es la de reemplazar el sistema de enfriamiento. Fin del programa.</li>

<li> Si el sistema de enfriamiento de las baterías funciona, entonces hay que llevar el coche al taller. Fin del programa.</li>

<li> Si el coche no va lento, entonces hay que preguntar si el coche acelera cuando no se pisa el acelerador.</li>

<li> Si el coche acelera cuando no se pisa el acelerador, entonces hay que preguntar si la prueba de conexión de la batería al motor es positiva.</li>

<li> Si el coche no acelera cuando no se pisa el acelerador, entonces hay que preguntar si el coche va a tirones.</li>

<li> Si el coche va a tirones, entonces hay que preguntar si la prueba de avería en la transmisión es positiva.</li>

<li> Si la prueba de avería en la transmisión es positiva, entonces la reparación sugerida es la de cambiar la transmisión. Fin del programa.</li>

<li> Si la prueba de avería en la transmisión es negativa, entonces hay que llevar el coche al taller. Fin del programa.</li>

<li> Si el coche no va a tirones, entonces hay que preguntar si se oye algún ruido.</li>

<li> Si no se oye ningún ruido, entonces hay que llevar el coche al taller. Fin del programa.</li>

<li> Si se oye algún ruido, entonces hay que preguntar si el ruido es constante.</li>

<li> Si el ruido es constante, entonces la reparación sugerida es que cambiar el motor. Fin del programa.</li>

<li> Si el ruido no es constante, entonces hay que llevar el coche al taller. Fin del programa.</li>


</ul>
