## Introducción a la programación de reglas con CLIPSPy

En cualquier programa que emplee CLIPSPy, en primer lugar hay que importar la biblioteca de CLIPS de Python.

```python
# Creamos el entorno de CLIPSPy para trabajar
entorno = clips.Environment()
```

Los elementos básicos de CLIPS son:

<ul>
<li>Números: reales (<i>float</i>) o enteros (<i>integer</i>). </li>
<li>Símbolos (<i>symbols</i>): cualquier secuencia de caracteres que no es un número. Se distingue entre mayúsculas y minúsculas.</li>

## Hechos en CLIPSPy

Un hecho es un dato o información que se almacena en la lista de hechos o base de datos. Cada hecho tiene un identificador que es el índice en la lista de hechos; se muestra de la forma f.XXX: por ejemplo, f.5 hace referencia al quinto hecho. Existen dos tipos de hechos:

+ Hechos ordenados: formados por un símbolo seguido por ninguno o por varios campos separados por espacios y delimitado todo entre paréntesis. El orden de los campos es importante.
+ Hechos no ordenados (`deftemplate`): asignan un nombre a cada campo del hecho. Cada campo tiene un nombre y el valor que toma ese nombre, todo ello delimitado entre paréntesis. Para definir los campos se requiere una plantilla o modelo que se tiene que definir antes de generar los hechos. Como los campos tienen el nombre y el valor, el orden no afecta al significado del hecho.



Suele ser mejor definir los hechos como hechos no ordenados, es decir, crear una plantilla y crear posteriormente los hechos. Sin embargo, hay un par de casos en los que es adecuado utilizar hechos ordenados:

•	Banderas o indicadores (flags). Son útiles por ejemplo para indicar que se ha alcanzado alguna situación.

•	Cuando los hechos contienen una sola casilla y el nombre de la casilla es sinónima del nombre de la relación.

En la siguiente celda de código se crea un hecho ordenado y se muestran los hechos con un bucle `for`.


```python
"""Crear una base de datos"""
# Hecho ordenado
# El símbolo cliente_hecho_ordenado equivale al nombre de la plantilla de un hecho no ordenado
hecho = entorno.assert_string('(cliente_hecho_ordenado "Luis Pérez López" 000001 9999999Z 110 209 352)')


# Se muestran los hechos. La base de datos se guarda en entorno.facts()
for f in entorno.facts():
  print('Ìndice hecho', f.index,':',f)
```

Los datos de un hecho ordenado se pueden extraer como una lista de datos. 

```python
# Todos los datos del hecho
datos_hecho = list(hecho)
print('Los datos del hecho son:')
print(datos_hecho)

# Se puede extraer un dato en particular. Por ejemplo, el segundo dato
print('El segundo dato del hecho es:')
print(hecho[1])
# O también así
print('El segundo dato del hecho es:')
print(datos_hecho[1])
```

Ahora, vamos a crear una serie de hechos no ordenados en CLIPSPy. La estructura de los hechos no ordenados es:

    variable_plantilla = """(deftemplate nombre_plantilla "Comentarios"
        (definición de los campos)           
         )"""

En la siguiente celda se muestra la generación de hechos no ordenados que llamamos
*cliente*. En primer lugar, se crea la plantilla:

```python
# Plantilla de hechos no ordenados. La variable de Python se llama plantilla_cliente
# Cada "slot" es un campo. Se utiliza TYPE para indicar el tipo del valor que toma el campo
plantilla_cliente = """(deftemplate cliente "Cliente de la empresa"
    (slot nombre (type STRING))
    (slot apellidos (type STRING))
    (slot numero_cliente (type INTEGER))
    (slot DNI (type SYMBOL)) )"""
```

A continuación, la plantilla se debe añadir al entorno con `build`. Para utilizar la plantilla, se debe recuperar con `find_template`.

```python
# La plantilla de hechos no ordenados se debe añadir al entorno
entorno.build(plantilla_cliente)

# Obtenemos la plantilla
plantilla_cliente_usar = entorno.find_template('cliente')
```

Ahora ya se pueden crear hechos no ordenados con la plantilla. Para ello empleamos la plantilla recuperada y `assert_fact`.

```python
# Dado el modelo anterior se puede crear un hecho o "fact" de tipo cliente
# Para ello se emplea el método assert_fact
# No hay que numerar los hechos o "facts" para que se creen
hecho = plantilla_cliente_usar.assert_fact(nombre='José',
                                      apellidos='Pérez López',
                                      numero_cliente=1,
                                      DNI=clips.Symbol('99999999Z'))

# Creamos un segundo hecho
hecho = plantilla_cliente_usar.assert_fact(nombre="María",
                                      apellidos="Tristán Luengo",
                                      numero_cliente=2,
                                      DNI=clips.Symbol('11111111M'))

# Se muestran los hechos. La base de datos se guarda en entorno.facts()
for f in entorno.facts():
  print('Índice hecho', f.index,':',f)
```

Los hechos se pueden eliminar mediante el método `retract` En la siguiente celda se muestra cómo eliminar un hecho.

```python
# Un hecho puede borrarse de la lista de hechos. Por ejemplo, queremos eliminar el último hecho
hecho.retract()

print('Hemos eliminado el último hecho')
for f in entorno.facts():
  print('Índice hecho', f.index,':',f)
```

Si damos un nombre particular a cada hecho, se puede eliminar un hecho en particular:

```python
# NOTA:
# Como se puede observar al ser el primer hecho de la celda anterior
# exactamente igual al hecho_1 de esta celda, no se duplican los hechos

# Si damos nombres diferentes a los hechos podemos eliminar uno en particular
hecho_1 = plantilla_cliente_usar.assert_fact(nombre='José',
                                      apellidos='Pérez López',
                                      numero_cliente=1,
                                      DNI=clips.Symbol('99999999Z'))

# Creamos un segundo hecho
hecho_2 = plantilla_cliente_usar.assert_fact(nombre="María",
                                      apellidos="Tristán Luengo",
                                      numero_cliente=2,
                                      DNI=clips.Symbol('11111111M'))

# Mostramos los hechos
print('Todos los hechos son:')
for f in entorno.facts():
  print('Ìndice hecho', f.index,':',f)

# Eliminamos el primer hecho no ordenado
hecho_1.retract()

# Mostramos los hechos
print('Hemos eliminado el hecho_1:')
for f in entorno.facts():
  print('Índice hecho', f.index,':',f)
```

Los hechos se pueden modificar mediante el método `modify_slots`. En la siguiente celda se muestra cómo cambiar el valor de dos de los campos.

```python
# Supongamos que nos hemos equivocado en el apellido y DNI de uno de los clientes
# Cambiamos el apellido y el DNI
hecho_2.modify_slots(apellidos="Tristán Carrascosa", DNI=clips.Symbol('22222222M'))
print('Hemos modificado un hecho:')
for f in entorno.facts():
  print('Índice hecho', f.index,':',f)
```

Los campos de los hechos pueden ser de tipo `multislot`. En este caso el campo podrá tener varios valores. Por ejemplo, en la plantilla cliente se podría añadir un campo que indicase los pedidos realizados por cada cliente:

```python
# Plantilla de hechos no ordenados con un campo multislot
plantilla_nueva_cliente = """(deftemplate nuevo_tipo_cliente "Cliente de la empresa"
    (slot nombre (type STRING))
    (slot apellidos (type STRING))
    (slot numero_cliente (type INTEGER))
    (slot DNI (type SYMBOL))
    (multislot numero_pedidos (type INTEGER)) )"""
```

```python
# La plantilla de hechos no ordenados se debe añadir al entorno
entorno.build(plantilla_nueva_cliente)

# Obtenemos la plantilla
plantilla_cliente_usar = entorno.find_template('nuevo_tipo_cliente')

# Nuevo hecho
hecho = plantilla_cliente_usar.assert_fact(nombre='José',
                                      apellidos='Pérez López',
                                      numero_cliente=1,
                                      DNI=clips.Symbol('99999999Z'),
                                      numero_pedidos=(110, 209, 352))

# Mostramos el hecho creado
for f in entorno.facts():
  print('Índice hecho', f.index,':',f)
```

En los hechos no ordenados también se pueden extraer los datos de los campos, para ello hay que utilizar el nombre del campo. Hay que tener en cuenta que la forma de extraer los datos no es la misma que la empleada en los hechos ordenados. 

```python
# Dato apellidos del hecho no ordenado 
print('El dato apellidos del hecho no ordenado es:')
print(hecho['apellidos'])
```

Si recorremos con un bucle for los hechos del entorno, hay que seleccionar si el hecho es ordenado o no ordenado para aplicar el método correcto.

Los hechos, tanto ordenados como no ordenados, se seleccionan mediante: `f.template.name`.

```python
# Recorremos los hechos para mostrar algunos datos
for f in entorno.facts():
    
    if f.template.name == 'nuevo_tipo_cliente':
        print('Los apellidos del cliente', f)
        print('son:', f['apellidos'])
    elif f.template.name == 'cliente':
        print('Los apellidos del cliente', f)
        print('son:', f['apellidos'])
    elif f.template.name == 'cliente_hecho_ordenado':
        print('El segundo dato del hecho ordenado es:')
        print(f[1])
```

## Reglas en CLIPSPy

Una regla se compone de un antecedente y de un consecuente; el primero está en la parte izquierda de la regla y el segundo en la parte derecha. El antecedente puede constar de varias condiciones; la regla se activa cuando se satisfacen los elementos condicionales del antecedente:

**Si** antecedente, **entonces** consecuente.

El motor de inferencia de CLIPS se encarga de decidir la activación de las reglas; en el caso de que se active más de una regla se utiliza una estrategia de resolución de conflictos para seleccionar la regla que debe activarse primero.

En la definición de reglas de CLIPS hay que tener en cuenta que:

•	El término Si se omite, está implícito.

•	El término entonces corresponde a =>.

•	Los elementos condicionales del antecedente están unidos de forma implícita por Y, por una conjunción.


Así, la sintaxis en CLIPSPy de una regla es:

    variable_regla = """
    (defrule nombre_regla "Comentarios"
        [<propiedades>]
        <antecendente>
        =>                        
       <acciones>
    )
"""

Si por ejemplo queremos definir una regla en lenguaje natural que indica: Si está lloviendo y no tengo paraguas, entonces me mojaré, escribiremos:

    regla_lluvia = """
    (defrule Lluvia "Regla sobre la lluvia denominada Lluvia"
        (EstaLloviendo si)
        (TengoParaguas no)
        =>                        
       (assert (MeMojare si))
    )
"""

La aplicación de reglas sigue las siguientes directrices:

•	Se dice que una regla se activa cuando se satisfacen los elementos condicionales de una regla.

•	Una regla puede activarse para distintos conjuntos de hechos (hay distintas razones para que la regla se active). En este caso, cada posible activación de la regla se llama instancia de una regla.

•	Una regla está desactivada si no está activada. Es decir, no existe ninguna instancia de una regla.

•	Una regla se dispara si se ha optado por realizar las acciones (consecuentes) de una regla, para poder dispararse tiene que estar activada.

•	Se conoce por agenda a la parte del sistema que contiene una lista de las reglas activadas. Una regla aparecerá tantas veces en la agenda como instancias existan de la regla.

•	Una estrategia de resolución de conflictos es el conjunto de criterios que permiten determinar, en el caso de que haya varias reglas en la agenda (varias reglas activas), qué regla debe dispararse.

•	Se llama prioridad de una regla a un valor numérico asociado a la regla que indica la importancia de la regla, y por tanto la prioridad con que debe ejecutarse. Cuanto mayor es el valor, mayor es la prioridad. Notar que todas las instancias de una regla tendrán la misma prioridad.

•	CLIPSPy permite establecer valores de prioridad entre los valores -10000 y +10000. Por defecto, todas las reglas tienen el valor de prioridad 0.

A continuación, vamos a estudiar cómo construir los antecedentes y consecuentes mediante ejemplos y ejercicios.


**Patrones: Literales.**

El antecedente de una regla consta de cero o más elementos condicionales. El elemento condicional más simple es un patrón. Cada patrón consta de una o varias restricciones sobre el valor de algunos atributos de un hecho (campos o casillas de la plantilla de un hecho no ordenado).

La restricción más básica que puede utilizarse en un elemento condicional es aquella en la que se exige un valor exacto para una casilla de un hecho; a este tipo de restricciones se les llama literales.

Vamos a construir una regla con un antecedente con un patrón literal. En primer lugar, en el siguiente ejercicio, construya una plantilla de hechos no ordenados y genere los hechos indicados.


**Ejercicio 1**

Construya una plantilla de hechos no ordenados llamada *Persona* que contenga los campos *nombre* y *apellido* de tipo *STRING*, *edad*  de tipo *INTEGER* y multislot *amigos* de tipo *STRING*.

A continuación, genere los hechos correspondientes a las siguientes personas:

•	Luis Pérez, de edad 30 años, y con amigos: José, Manolo y Daniel.

•	Daniel Rubio, de edad 25 años y con amigos: José, Daniel y María.

•	José Pérez, de edad 32 años y con amigos: Daniel y Luis.

•	Manolo Palma, de edad 30 años y con amigos: María y Daniel.

•	María Rubio, de edad 30 años y con amigos Daniel, José y Manolo.

**Cree un nuevo entorno de CLIPS para estos hechos y las reglas llamado entorno_ejercicios.**


```python
entorno_ejercicios = clips.Environment()

persona = """(deftemplate Persona "Persona"
    (slot nombre (type STRING))
    (slot apellido (type STRING))
    (slot edad (type INTEGER))
    (multislot amigos (type STRING)) )"""

entorno_ejercicios.build(persona)

plantilla_persona_usar = entorno_ejercicios.find_template('Persona')
```

```python
hecho1 = plantilla_persona_usar.assert_fact(nombre='Luis',
                                      apellido='Pérez',
                                      edad=30,
                                      amigos=('José', 'Manolo', 'Daniel'))
hecho2 = plantilla_persona_usar.assert_fact(nombre='Daniel',
                                      apellido='Rubio',
                                      edad=25,
                                      amigos=('José', 'Daniel', 'María'))
hecho = plantilla_persona_usar.assert_fact(nombre='José',
                                      apellido='Pérez',
                                      edad=32,
                                      amigos=('Daniel', 'Luis'))
hecho3 = plantilla_persona_usar.assert_fact(nombre='Manolo',
                                      apellido='Palma',
                                      edad=30,
                                      amigos=('María', 'Daniel'))
hecho4 = plantilla_persona_usar.assert_fact(nombre='María',
                                      apellido='Rubio',
                                      edad=30,
                                      amigos=('Daniel', 'José', 'Manolo'))
```

Ahora se muestra un regla que genera el hecho ordenado *Persona_con_apellido Pérez* para cada persona cuyo apellido sea Pérez.

Para que las reglas se puedan ejecutar hay que añadir la regla al entorno con el método `build()`.

```python
# Definimos una regla
regla_hecho_ordenado_Perez = """
(defrule hecho_ordenado_Perez "Esta regla genera un hecho ordenado si se cumple una condición"
    ( Persona (apellido "Pérez") )
    =>
    ( assert (Persona_con_apellido Pérez) )
)
"""

# La regla se añade al entorno
entorno_ejercicios.build(regla_hecho_ordenado_Perez)

# Con el siguiente código podemos comprobar las reglas existentes en el entorno
for rule in entorno_ejercicios.rules():
    print(rule)
```

Para ejecutar las reglas hay que ejecutar el motor de inferencia con el método `run()`.

```python
# Se ejecuta el motor de inferencia: comprueba las reglas que se actiavn y procede a ejecutarlas en orden
# En este caso sólo hay una regla
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Índice hecho', f.index,':',f)
```

**Patrones: variables unicampo.**

Una variable unicampo captura el valor de un campo de un hecho. El nombre de una variable unicampo empieza siempre por "?". Si aparece la misma variable en más de un elemento condicional, la primera ocurrencia captura el valor, y en las restantes ocurrencias se sustituye el valor capturado por la variable.

En la siguiente celda de código se modifica la regla anterior para mostrar el nombre de cada una de las personas que se apellidan Pérez.


```python
# Definimos la regla
regla_hecho_ordenado_Perez_nombre = """
(defrule hecho_ordenado_Perez_nombre "Esta regla genera un hecho ordenado si se cumple una condición"
    ( Persona (apellido "Pérez") (nombre ?x) )
    =>
    ( assert (Persona_con_nombre_apellidos ?x Pérez) )
)
"""
# EL nombre lo guarda -> Se crea un hecho con con cada nombre, ya que guardo el nomber y lo guardo en el hecho


# La regla se añade al entorno
entorno_ejercicios.build(regla_hecho_ordenado_Perez_nombre)

# Con el siguiente código podemos comprobar las reglas existentes en el entorno
for rule in entorno_ejercicios.rules():
    print(rule)

# Se ejecuta el motor de inferencia: comprueba las reglas que se activan y procede a ejecutarlas en orden
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)
```

**Patrones: comodines.**

Un comodín representa cualquier valor que se encuentre almacenado en una casilla. Se diferencia de las variables unicampo en que no guardan el valor de la casilla en una variable. Se distinguen dos tipos de comodines:

•	Comodín unicampo, representado por el símbolo ?, sustituye cualquier valor de una casilla, y exactamente uno, no por varios valores. 

•	Comodín multicampo, representado por el símbolo $?, sustituye a cero o más valores de una casilla.

A continuación, veremos la utilidad del **comodín unicampo**. Supongamos que se desea mostrar los amigos de una determinada persona. Se podría ejecutar una regla que buscase a las personas amigas de “?n” que tienen tres amigos y en los que “?x” aparece como tercer amigo.


```python
# Definimos la regla
regla_busca_personas_tercer_amigo = """
(defrule Busca_personas_tercer_amigo "Busca las personas con un tercer amigo"
  ( Persona (nombre ?n) (amigos ? ? ?x) )
  =>
  ( assert (Persona_nombre_tercer_amigo ?n ?x) )
)
"""
```

Imaginemos que queremos mostrar las personas amigas de “?n” en las que “?x” aparece como el último amigo de la lista; si hubiese M posibles valores de la casilla multicampo (multislot) habría que escribir M reglas.
```python
# Definimos la regla
regla_busca_personas_ultimo_amigo = """
(defrule Busca_personas_ultimo_amigo "Busca las personas y muestra su último amigo"
  ( Persona (nombre ?n) (amigos $? ?x) )
  =>
  ( assert (Persona_nombre_ultimo_amigo ?n ?x) )
)
"""

# Añadimos la regla al entorno
entorno_ejercicios.build(regla_busca_personas_ultimo_amigo)

# Con el siguiente código podemos comprobar las reglas existentes en el entorno
for rule in entorno_ejercicios.rules():
    print(rule)

# Se ejecuta el motor de inferencia: comprueba las reglas que se actiavn y procede a ejecutarlas en orden
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)
```

**Patrones: variables multicampo.**

Como ha podido apreciar, la diferencia sustancial entre el comodín “?” y una variable unicampo “?x” consiste en que:

•	Con un comodín se hace la sustitución de un valor, el valor que sea que exista en los hechos

•	Con una variable unicampo, un valor determinado se asocia a la variable “?x”.

Con el comodín multicampo *?* se pueden crear variables multicampo con la forma $?.  Vamos a ver un ejemplo de regla que permite mostrar los amigos que están antes de un nombre de amigo que "buscaremos" con un hecho ordenado (por ejemplo Daniel, observe que Daniel tiene un amigo que a su vez se llama Daniel).

```python
# Definimos la regla
regla_busca_amigos_antes = """
(defrule Busca_amigos_antes "Busca los amigos situados en la lista antes que otro"
  ( Nombre-Amigo-Buscar ?x )
  ( Persona (nombre ?n) (amigos $?antes ?x $?) )
  =>
  ( assert (Persona_nombre_amigo ?n $?antes ?x))
)
"""

# Definimos el hecho ordenado
entorno_ejercicios.assert_string('(Nombre-Amigo-Buscar "Daniel")')


# Añadimos la regla al entorno
entorno_ejercicios.build(regla_busca_amigos_antes)

# Con el siguiente código podemos comprobar las reglas existentes en el entorno
for rule in entorno_ejercicios.rules():
    print(rule)

# Se ejecuta el motor de inferencia: comprueba las reglas que se activan y procede a ejecutarlas en orden
entorno_ejercicios.run()

# Mostramos los hechos
for f in entorno_ejercicios.facts():
  print('Ìndice hecho', f.index,':',f)
```

**Conectivas en una casilla.**

Hasta ahora se han visto restricciones de una casilla establecidos en términos de literales, comodines y variables. CLIPSPy permite establecer relaciones entre estos tipos de restricciones mediante (restricciones) conectivas. A saber:

•	Restricción **no**. Se denota por **~** y se satisface si la restricción que le sigue no se satisface.

•	Restricción **y**. Se denota por **&** y se satisface si las restricciones adyacentes se satisfacen.

•	Restricción **o**. Se denota por **|** y se satisface si alguna de las restricciones adyacentes se satisfacen.

En el caso de existir varias conectivas en una casilla, la restricción ~ sería la primera en realizarse, seguida de las restricciones &, y  finalizando con las restricciones |. En caso de empate, la evaluación de las restricciones se hace en el orden de aparición (de izquierda a derecha).
Los siguientes ejemplos muestran el uso de estos conectores.

Para mostrar sólo las nuevas reglas, en primer lugar borramos el entorno, con el método `clear()`.

```python
# Regla que emplea la conectiva No (~)
regla_semaforo_1 = """
(defrule regla_semaforo_1 "Regla 1 del semáforo"
  ( semaforo ~verde )
  =>
  ( assert (Cruzar No_puedo) )
)
"""
```
