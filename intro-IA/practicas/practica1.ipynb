**La función aplica el descuento al precio de un producto**

## *Ejercicio1-2 ❗

Escribe un programa que haga las siguientes tres cosas:

1. Almacene en una lista llamada `precios` los siguientes precios: 50, 75, 46, 22, 80, 65, 8.
2. Que obtenga el mayor y el menor de los valores de la lista `precios`, recorriendo la lista una sola vez. Utiliza las variables `min` y `max` para ir guardando cada valor.
3. Que imprima por pantalla el valor de las variables `min` y `max`, que deberían representar el menor y mayor valor de la lista

```python
precios = [50, 75, 46, 22, 80, 65, 8]
min = precios[0]
max = precios[0]
for i in precios:
  if i < min:
    min = i
  if i > max:
    max = i
print(max, min)
```

## *Ejercicio1-3 ❗

Crea una función `caracteres_unicos(cadena1, cadena2)` con dos parámetros, la cuál va a recibir dos cadenas (`cadena1` y `cadena2`).  Esta función deberá utilizar la funcionalidad de las listas y los conjuntos, para devolver el conjunto de carácteres únicos entre `cadena1` y `cadena2`.

Por lo tanto, una ejecución como `caracteres_unicos('hola', 'caracola')` debería devolver los siguientes caracteres únicos: `{'r', 'h', 'l', 'a', 'o', 'c'}`



Ten en cuenta, que puedes transformar una cadena en una lista de los caracteres que la componen mediante `list(cadena)`, mira el siguiente ejemplo:

```python
print(set('abcc'))
```

Ten en cuenta también, que los conjuntos disponen del método `union()`, tal que `set1.union(set2)` nos devolvería el conjunto de elementos únicos en los dos conjuntos.

```python
def caracteres_unicos(cadena1, cadena2):
  return (set(cadena1).union(set(cadena2)))

caracteres_unicos("hola", "caracola")
```
